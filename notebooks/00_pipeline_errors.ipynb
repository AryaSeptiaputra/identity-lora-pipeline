# 00 — Error Pipeline

Definisi error milik `identity-lora-pipeline`. Notebook ini dimuat oleh `01_human_detection_and_cropping.ipynb`, `02_face_detection_and_cropping.ipynb`, dan `03_image_captioning.ipynb` lewat `%run`, jadi semua notebook pipeline memakai kelas error yang sama.

## 1. Error pipeline

```python
class ImageReadError(Exception):
    """Gambar mentah gagal dibaca."""


class DetectionError(Exception):
    """Model deteksi gagal dijalankan pada gambar."""


class NoDetectionError(Exception):
    """Tidak ada box "person" yang terdeteksi pada gambar."""


class MultipleDetectionError(Exception):
    """Lebih dari satu box "person" terdeteksi pada gambar (di luar cakupan MVP)."""


class CropTooSmallError(Exception):
    """Sisi terpanjang hasil crop di bawah ukuran minimum."""


class CropSaveError(Exception):
    """Hasil crop gagal disimpan ke berkas."""



class FaceDetectionError(Exception):
    """Model deteksi wajah gagal dijalankan pada gambar."""


class NoFaceDetectedError(Exception):
    """Tidak ada wajah yang terdeteksi pada gambar."""


class MultipleFaceDetectedError(Exception):
    """Lebih dari satu wajah terdeteksi pada gambar (di luar cakupan MVP)."""


class PoseDetectionError(Exception):
    """Model deteksi pose gagal dijalankan pada gambar."""


class NoPersonPoseDetectedError(Exception):
    """Tidak ada orang yang terdeteksi oleh model pose pada gambar."""


class MultiplePersonPoseDetectedError(Exception):
    """Lebih dari satu orang terdeteksi oleh model pose pada gambar (di luar cakupan MVP)."""


class InsufficientKeypointsError(Exception):
    """Keypoint grup upper/lower body tidak cukup untuk membentuk bounding box."""


class GPUNotAvailableError(Exception):
    """CUDAExecutionProvider tidak tersedia saat GPU wajib (K9, DWPose)."""


class CaptionGenerationError(Exception):
    """Model captioning gagal dijalankan pada batch gambar."""
```
